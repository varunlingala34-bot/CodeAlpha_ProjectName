```python
# ============================================================
# CODEALPHA DATA SCIENCE INTERNSHIP
# TASK 2 - UNEMPLOYMENT ANALYSIS IN INDIA
# COVID-19 IMPACT ANALYSIS
# ============================================================

# -----------------------------
# 1. UPLOAD DATASET
# -----------------------------

from google.colab import files
import zipfile
import os

print("Please upload the Kaggle archive.zip file")
uploaded = files.upload()

zip_name = list(uploaded.keys())[0]

# Create data folder
os.makedirs("data", exist_ok=True)

# Extract ZIP
with zipfile.ZipFile(zip_name, "r") as zip_ref:
    zip_ref.extractall("data")

print("\nFiles extracted successfully:")
print(os.listdir("data"))


# -----------------------------
# 2. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (11, 5)


# -----------------------------
# 3. LOAD DATASET
# -----------------------------

df = pd.read_csv("data/Unemployment in India.csv")

print("\nRows and Columns:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())


# -----------------------------
# 4. INSPECT DATA
# -----------------------------

print("\nOriginal Column Names:")
print(list(df.columns))

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nCompletely Empty Rows:")
print(df.isnull().all(axis=1).sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# -----------------------------
# 5. CLEAN DATA
# -----------------------------

# Remove spaces from column names
df.columns = df.columns.str.strip()

# Rename columns
df = df.rename(columns={
    "Estimated Unemployment Rate (%)": "Unemployment_Rate",
    "Estimated Employed": "Employed",
    "Estimated Labour Participation Rate (%)": "Labour_Participation"
})

# Remove completely empty rows
df = df.dropna(how="all")

# Remove duplicate rows
df = df.drop_duplicates()

# Remove unnecessary spaces from text columns
for col in ["Region", "Frequency", "Area"]:
    df[col] = df[col].str.strip()

# Convert Date column
df["Date"] = pd.to_datetime(
    df["Date"].str.strip(),
    format="%d-%m-%Y"
)

# Create additional date columns
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.strftime("%b")

# Classify COVID period
# April 2020 onward is treated as the COVID period
df["Period"] = np.where(
    df["Date"] < "2020-04-01",
    "Before COVID",
    "During COVID"
)

# Sort data
df = df.sort_values("Date").reset_index(drop=True)

print("\n================ CLEANING RESULTS ================")
print("Rows and Columns after cleaning:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print(
    "Date range:",
    df["Date"].min().date(),
    "to",
    df["Date"].max().date()
)
print("Number of states/regions:", df["Region"].nunique())

print("\nCleaned Columns:")
print(list(df.columns))

print("\nFirst 5 cleaned rows:")
display(df.head())


# -----------------------------
# 6. EXPLORATORY DATA ANALYSIS
# -----------------------------

print("\n================ STATISTICAL SUMMARY ================")
display(df.describe().round(2))

print("\nAreas:")
print(df["Area"].unique())

print("\nFrequency:")
print(df["Frequency"].unique())

print("\nRows by Period:")
print(df["Period"].value_counts())


# -----------------------------
# 7. AVERAGE UNEMPLOYMENT TREND
# -----------------------------

monthly = df.groupby("Date")["Unemployment_Rate"].mean()

plt.figure(figsize=(12, 5))

plt.plot(
    monthly.index,
    monthly.values,
    marker="o",
    linewidth=2
)

plt.axvline(
    pd.Timestamp("2020-03-25"),
    linestyle="--",
    linewidth=2,
    label="Lockdown: 25 Mar 2020"
)

plt.title(
    "Average Unemployment Rate in India"
)

plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()


# -----------------------------
# 8. RURAL VS URBAN ANALYSIS
# -----------------------------

area_trend = (
    df.groupby(["Date", "Area"])["Unemployment_Rate"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=area_trend,
    x="Date",
    y="Unemployment_Rate",
    hue="Area",
    marker="o",
    linewidth=2
)

plt.axvline(
    pd.Timestamp("2020-03-25"),
    linestyle="--",
    linewidth=2
)

plt.title(
    "Unemployment Rate: Rural vs Urban"
)

plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -----------------------------
# 9. STATE-WISE AVERAGE
# -----------------------------

state_avg = (
    df.groupby("Region")["Unemployment_Rate"]
    .mean()
    .sort_values(ascending=False)
)

top_states = state_avg.head(10)

plt.figure(figsize=(11, 6))

sns.barplot(
    x=top_states.values,
    y=top_states.index
)

plt.title(
    "Top 10 States by Average Unemployment Rate"
)

plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("State")

plt.tight_layout()
plt.show()

print("\nTop 10 States:")
display(top_states.round(2))


# -----------------------------
# 10. STATE-WISE HEATMAP
# -----------------------------

heat = df.pivot_table(
    index="Region",
    columns="Date",
    values="Unemployment_Rate",
    aggfunc="mean"
)

heat.columns = [
    date.strftime("%b-%y")
    for date in heat.columns
]

plt.figure(figsize=(14, 10))

sns.heatmap(
    heat,
    linewidths=0.3
)

plt.title(
    "Unemployment Rate (%) by State and Month"
)

plt.xlabel("Month")
plt.ylabel("State")

plt.tight_layout()
plt.show()


# -----------------------------
# 11. COVID IMPACT BY STATE
# -----------------------------

covid = df.pivot_table(
    index="Region",
    columns="Period",
    values="Unemployment_Rate",
    aggfunc="mean"
)

covid["Increase"] = (
    covid["During COVID"]
    - covid["Before COVID"]
)

covid = covid.sort_values(
    "Increase",
    ascending=False
)

print("\nStates with the Largest Increase:")
display(covid.head(10).round(2))


# -----------------------------
# 12. COVID IMPACT GRAPH
# -----------------------------

top_covid = covid.head(10)

plt.figure(figsize=(11, 6))

sns.barplot(
    x=top_covid["Increase"].values,
    y=top_covid.index
)

plt.title(
    "Top 10 States: Increase in Unemployment During COVID-19"
)

plt.xlabel(
    "Increase in Unemployment Rate (percentage points)"
)

plt.ylabel("State")

plt.tight_layout()
plt.show()


# -----------------------------
# 13. OVERALL COVID COMPARISON
# -----------------------------

overall = (
    df.groupby("Period")["Unemployment_Rate"]
    .mean()
)

print("\n================ OVERALL COVID COMPARISON ================")

print("\nAverage Unemployment Rate:")
display(overall.round(2))

increase = (
    overall["During COVID"]
    - overall["Before COVID"]
)

print(
    "\nIncrease in percentage points:",
    round(increase, 2)
)


# -----------------------------
# 14. RURAL VS URBAN COVID IMPACT
# -----------------------------

by_area = df.pivot_table(
    index="Area",
    columns="Period",
    values="Unemployment_Rate",
    aggfunc="mean"
)

print("\n================ RURAL VS URBAN COVID IMPACT ================")
display(by_area.round(2))

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Area",
    y="Unemployment_Rate",
    hue="Period",
    errorbar=None
)

plt.title(
    "Average Unemployment Rate: Before vs During COVID-19"
)

plt.xlabel("Area")
plt.ylabel("Unemployment Rate (%)")

plt.tight_layout()
plt.show()


# -----------------------------
# 15. EMPLOYMENT TREND
# -----------------------------

trend2 = df.groupby("Date").agg(
    Employed=("Employed", "sum"),
    Labour_Participation=("Labour_Participation", "mean")
)

plt.figure(figsize=(12, 5))

plt.plot(
    trend2.index,
    trend2["Employed"] / 1e6,
    marker="o",
    linewidth=2
)

plt.title(
    "Total Estimated Employment Over Time"
)

plt.xlabel("Date")
plt.ylabel("Estimated Employed (Millions)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -----------------------------
# 16. LABOUR PARTICIPATION TREND
# -----------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    trend2.index,
    trend2["Labour_Participation"],
    marker="o",
    linewidth=2
)

plt.title(
    "Average Labour Participation Rate Over Time"
)

plt.xlabel("Date")
plt.ylabel("Labour Participation Rate (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -----------------------------
# 17. PRE-COVID MONTHLY PATTERN
# -----------------------------

pre_covid = df[
    df["Period"] == "Before COVID"
]

season = (
    pre_covid.groupby("Date")["Unemployment_Rate"]
    .mean()
)

plt.figure(figsize=(12, 5))

sns.barplot(
    x=season.index.strftime("%b-%y"),
    y=season.values
)

plt.title(
    "Average Unemployment Rate by Month Before COVID-19"
)

plt.xlabel("Month")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(
    "\nPre-COVID unemployment range:",
    round(season.min(), 2),
    "to",
    round(season.max(), 2),
    "%"
)


# -----------------------------
# 18. CORRELATION ANALYSIS
# -----------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Labour_Participation",
    y="Unemployment_Rate",
    hue="Period",
    alpha=0.7
)

plt.title(
    "Unemployment Rate vs Labour Participation"
)

plt.xlabel(
    "Labour Participation Rate (%)"
)

plt.ylabel(
    "Unemployment Rate (%)"
)

plt.tight_layout()
plt.show()

print("\n================ CORRELATION MATRIX ================")

correlation = df[
    [
        "Unemployment_Rate",
        "Employed",
        "Labour_Participation"
    ]
].corr()

display(correlation.round(2))


# -----------------------------
# 19. AUTOMATIC KEY INSIGHTS
# -----------------------------

peak_date = monthly.idxmax()

before_avg = overall["Before COVID"]
during_avg = overall["During COVID"]

worst_state = covid.index[0]

top_state_overall = state_avg.index[0]

rural_before = by_area.loc[
    "Rural",
    "Before COVID"
]

rural_during = by_area.loc[
    "Rural",
    "During COVID"
]

urban_before = by_area.loc[
    "Urban",
    "Before COVID"
]

urban_during = by_area.loc[
    "Urban",
    "During COVID"
]

pre_monthly = monthly[
    monthly.index < "2020-04-01"
]


print("\n")
print("=" * 70)
print("                 KEY INSIGHTS")
print("=" * 70)

print(
    f"\n1. Average unemployment was "
    f"{before_avg:.1f}% before COVID and "
    f"{during_avg:.1f}% during COVID."
)

print(
    f"\n2. The highest average unemployment "
    f"occurred in {peak_date.strftime('%B %Y')} "
    f"at {monthly.max():.1f}%."
)

print(
    f"\n3. {worst_state} recorded the largest "
    f"increase between the two periods: "
    f"+{covid['Increase'].iloc[0]:.1f} "
    f"percentage points."
)

print(
    f"\n4. Urban unemployment changed from "
    f"{urban_before:.1f}% to "
    f"{urban_during:.1f}%."
)

print(
    f"   Rural unemployment changed from "
    f"{rural_before:.1f}% to "
    f"{rural_during:.1f}%."
)

print(
    f"\n5. The state with the highest average "
    f"unemployment over the dataset period "
    f"was {top_state_overall} "
    f"({state_avg.iloc[0]:.1f}%)."
)

print(
    f"\n6. Before COVID, monthly unemployment "
    f"ranged from {pre_monthly.min():.1f}% "
    f"to {pre_monthly.max():.1f}%."
)

print(
    "\n7. The COVID period produced a much "
    "larger change than the normal "
    "month-to-month variation observed "
    "before COVID."
)

print(
    f"\n8. By June 2020, the unemployment "
    f"rate had moved to {monthly.iloc[-1]:.1f}%."
)

print("\n" + "=" * 70)


# -----------------------------
# 20. OPTIONAL RECOVERY ANALYSIS
# -----------------------------

try:

    df2 = pd.read_csv(
        "data/Unemployment_Rate_upto_11_2020.csv"
    )

    df2.columns = df2.columns.str.strip()

    df2 = df2.rename(columns={
        "Estimated Unemployment Rate (%)":
            "Unemployment_Rate",

        "Estimated Employed":
            "Employed",

        "Estimated Labour Participation Rate (%)":
            "Labour_Participation"
    })

    df2["Date"] = pd.to_datetime(
        df2["Date"].str.strip(),
        format="%d-%m-%Y"
    )

    recovery = (
        df2.groupby("Date")["Unemployment_Rate"]
        .mean()
    )

    plt.figure(figsize=(12, 5))

    plt.plot(
        recovery.index,
        recovery.values,
        marker="o",
        linewidth=2
    )

    plt.axvline(
        pd.Timestamp("2020-03-25"),
        linestyle="--",
        linewidth=2,
        label="Lockdown: 25 Mar 2020"
    )

    plt.title(
        "Average Unemployment Rate: Jan-Oct 2020"
    )

    plt.xlabel("Date")
    plt.ylabel("Unemployment Rate (%)")

    plt.xticks(rotation=45)
    plt.legend()

    plt.tight_layout()
    plt.show()

except Exception as e:

    print(
        "\nOptional recovery dataset could not be loaded."
    )

    print("Error:", e)


# -----------------------------
# 21. FINAL CONCLUSION
# -----------------------------

print("\n")
print("=" * 70)
print("                       CONCLUSION")
print("=" * 70)

print("""
COVID-19 had a significant impact on unemployment
in India during the period covered by this dataset.

The unemployment rate increased sharply during the
COVID-19 period, with the largest changes occurring
around the lockdown period.

The analysis also shows differences between rural
and urban areas and between individual states.

Employment and labour participation also changed
during the same period, indicating a major disruption
in the labour market.

The COVID-19 shock was much larger than the normal
month-to-month variation observed before the pandemic.

However, this dataset covers only a limited period
from May 2019 to June 2020, so it is not sufficient
to establish long-term seasonal trends.

The national averages used in this project are simple
averages across available state observations and are
not population-weighted.
""")

print("=" * 70)
print("              TASK 2 ANALYSIS COMPLETED")
print("=" * 70)
```
